```python
# -------- Disjoint Set Union --------
class DSU:
    def __init__(self, n):
        self.parent = list(range(n))

    def find(self, x):
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])
        return self.parent[x]

    def union(self, u, v):
        pu = self.find(u)
        pv = self.find(v)
        if pu != pv:
            self.parent[pv] = pu

    def connected(self, u, v):
        return self.find(u) == self.find(v)


print("\n--- Network Connectivity Using DSU ---")

# -------- INPUT --------
n = int(input("Enter number of servers: "))

dsu = DSU(n)

m = int(input("Enter number of connections: "))

print("Enter connections (u v):")
for _ in range(m):
    u, v = map(int, input().split())
    dsu.union(u, v)

q = int(input("Enter number of connectivity queries: "))

print("Enter queries (u v):")
for _ in range(q):
    u, v = map(int, input().split())
    if dsu.connected(u, v):
        print("YES")
    else:
        print("NO")
```
