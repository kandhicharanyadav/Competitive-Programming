**Problem - 1 :**

```python
# ---------- INPUT ----------
n, m = map(int, input().split())

edges = []
for _ in range(m):
    u, v, w = map(int, input().split())
    edges.append((u, v, w))

S = int(input())

# ---------- BELLMAN-FORD ----------
INF = float('inf')
dist = [INF] * n
dist[S] = 0

# Relax edges N-1 times
for _ in range(n - 1):
    for u, v, w in edges:
        if dist[u] != INF and dist[u] + w < dist[v]:
            dist[v] = dist[u] + w

# Check negative cycle
negative_cycle = False
for u, v, w in edges:
    if dist[u] != INF and dist[u] + w < dist[v]:
        negative_cycle = True
        break

# ---------- OUTPUT ----------
if negative_cycle:
    print("NEGATIVE CYCLE")
else:
    print(*dist)
```
