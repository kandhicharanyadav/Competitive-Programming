**Problem - 1 :**

```python
from collections import deque
import sys
sys.setrecursionlimit(10**7)

# ---------- INPUT ----------
n, m = map(int, input().split())

adj = [[] for _ in range(n)]
for _ in range(m):
    u, v = map(int, input().split())
    adj[u].append(v)
    adj[v].append(u)

S, D = map(int, input().split())

# ---------- BFS (Shortest Path) ----------
def bfs_shortest_path():
    dist = [-1] * n
    q = deque([S])
    dist[S] = 0

    while q:
        node = q.popleft()
        if node == D:
            return dist[node]

        for nei in adj[node]:
            if dist[nei] == -1:
                dist[nei] = dist[node] + 1
                q.append(nei)

    return -1

# ---------- DFS (Cycle Detection) ----------
visited = [False] * n

def dfs(node, parent):
    visited[node] = True
    for nei in adj[node]:
        if not visited[nei]:
            if dfs(nei, node):
                return True
        elif nei != parent:
            return True
    return False

def has_cycle():
    for i in range(n):
        if not visited[i]:
            if dfs(i, -1):
                return True
    return False

# ---------- OUTPUT ----------
print(bfs_shortest_path())
print("YES" if has_cycle() else "NO")
```
