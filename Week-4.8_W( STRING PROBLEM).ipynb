**Problem - 1 :**

```python
# -------- Trie Node --------
class TrieNode:
    def __init__(self):
        self.children = {}
        self.end = False


# -------- Trie --------
class Trie:
    def __init__(self):
        self.root = TrieNode()

    # Insert contact
    def insert(self, word):
        node = self.root
        for ch in word:
            if ch not in node.children:
                node.children[ch] = TrieNode()
            node = node.children[ch]
        node.end = True

    # Search full word
    def search(self, word):
        node = self.root
        for ch in word:
            if ch not in node.children:
                return False
            node = node.children[ch]
        return node.end

    # Prefix check
    def startsWith(self, prefix):
        node = self.root
        for ch in prefix:
            if ch not in node.children:
                return False
            node = node.children[ch]
        return True


print("\n--- Mobile Contact Search Using Trie ---")

n = int(input("Enter number of contacts: "))

contacts = input(
    "Enter contact names (space-separated): "
).split()

trie = Trie()

# Insert contacts
for name in contacts:
    trie.insert(name)

# Search operation
name = input("Enter contact name to search: ")
print("Contact found →", trie.search(name))

# Search operation 2
name = input("Enter another contact name to search: ")
print("Contact found →", trie.search(name))

# Prefix queries
prefix = input("Enter prefix to check contacts: ")
print("Prefix exists →", trie.startsWith(prefix))

prefix = input("Enter another prefix to check contacts: ")
print("Prefix exists →", trie.startsWith(prefix))
```
