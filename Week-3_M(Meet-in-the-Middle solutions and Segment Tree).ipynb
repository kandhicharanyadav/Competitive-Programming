```python
from bisect import bisect_left

def get_subset_sums(arr):
    n = len(arr)
    sums = []
    for mask in range(1 << n):
        total = 0
        for i in range(n):
            if mask & (1 << i):
                total += arr[i]
        sums.append(total)
    return sums


def subset_sum_mitm(arr, target):
    n = len(arr)
    mid = n // 2

    left = arr[:mid]
    right = arr[mid:]

    left_sums = get_subset_sums(left)
    right_sums = get_subset_sums(right)

    right_sums.sort()

    for s in left_sums:
        need = target - s
        idx = bisect_left(right_sums, need)
        if idx < len(right_sums) and right_sums[idx] == need:
            return True

    return False


# Main input handling
T = int(input())
for _ in range(T):
    N, S = map(int, input().split())
    arr = list(map(int, input().split()))

    if subset_sum_mitm(arr, S):
        print("YES")
    else:
        print("NO")
```
